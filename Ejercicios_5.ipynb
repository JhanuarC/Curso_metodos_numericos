- **3c)** Aproxime $h$ con `bisect` sobre $[0,2]$, con `xtol=1e-12`, y reporte el valor absoluto de la diferencia con **3b)**.

```python
from scipy.optimize import newton , bisect
from math import pi

f = lambda h: h**3 - 3*h**2 + 2.4

#Respuesta 3b
raiz_newton = newton(f,x0=1,tol =1e-10)
#Respuesta 3c
raiz_bisect = bisect(f,a=0,b = 2,xtol=1e-12)

#Respuesta 3d
f_2 = lambda h: pi*h**2*(3 - h)/3
vol_sumer = round(f_2(raiz_bisect),4)
valor_bool = False

if  vol_sumer == round(((4*pi/3)*0.6),4):
    valor_bool = True
else:
    pass



R["3a"] = ((f(0)),(f(1)),(f(2)))   # (f(0), f(1), f(2))
R["3b"] = raiz_newton   # h por newton
R["3c"] = (raiz_bisect,abs(raiz_newton - raiz_bisect))   # (h por bisect, diferencia)
R["3d"] = (vol_sumer,valor_bool)   # (raices, volumen sumergido)

```

## Ejercicio 4 (20 puntos)

Sea

$$A=\begin{bmatrix} 8 & -2 & 1 & 0\\ -2 & 9 & -3 & 1\\ 1 & -3 & 10 & -2\\ 0 & 1 & -2 & 7\end{bmatrix},
\qquad \mathbf{b}=A\,[1,1,1,1]^{T}.$$

Sin usar ciclos.

- **4a)** Reporte $\|A\|_\infty$, $\det(A)$, $\kappa_\infty(A)$ y el valor booleano de la dominancia diagonal estricta de $A$.
- **4b)** Reporte los coeficientes del polinomio característico de $T_j$ con `np.poly`, y los radios espectrales $\rho(T_j)$ y $\rho(T_{gs})$.
- **4c)** Reporte $\rho(T_\omega)$ para $\omega=0.9,\,1.0,\,1.1,\,1.2$ y cuál de los cuatro valores da el menor radio espectral.
- **4d)** Resuelva $A\mathbf{x}=\mathbf{b}$ con `scipy.linalg.solve` y reporte $\mathbf{x}$ junto con la norma infinito del residual.

```python
import numpy as np
from scipy.linalg import solve
A = np.array([[ 8, -2,  1,  0],
              [-2,  9, -3,  1],
              [ 1, -3, 10, -2],
              [ 0,  1, -2,  7]], dtype=float)
b = A @ np.ones(4)

#4a

norm_inf_A = np.linalg.norm(A,np.inf)

det_A = np.linalg.det(A)

cond_inf_A = np.linalg.cond(A, np.inf)

diag = np.abs(np.diag(A))
sum_off_diag = np.sum(np.abs(A), axis=1) - diag
es_diagonal_dominante = bool(np.all(diag > sum_off_diag))


##Descomposicion A = D - L - U 4b

d = np.diag(np.diag(A))
l = -np.tril(A,k = -1)
u = -np.triu(A,k = 1)

Tj = np.linalg.inv(d) @ (l + u)
coe_poly_Tj = np.poly(Tj)


Tgs = np.linalg.inv(d-l) @ u

#Radios espectrales
r_esp_Tj = max(abs(np.linalg.eigvals(Tj)))
r_esp_Tgs = max(abs(np.linalg.eigvals(Tgs)))

#4c
w_val = [0.9,1.0,1.1,1.2]

w_optimo = None
rho_omega_min = float('inf')

for w in w_val:
    m = d - w*l
    n = (1-w)*d + w*u
    Tw = np.linalg.inv(m) @ n
    r_esp_Tw = max(abs(np.linalg.eigvals(Tw)))
    
    if r_esp_Tw <= rho_omega_min:
        w_optimo = w
        rho_omega_min = r_esp_Tw

#4d
x = solve(A, b)
r = A @ x - b


R["4a"] = (norm_inf_A, det_A, cond_inf_A, es_diagonal_dominante)   # (norm_inf_A, det_A, cond_inf_A, es_diagonal_dominante)
R["4b"] = (coe_poly_Tj,r_esp_Tj,r_esp_Tgs)   # (coef_poly_Tj, rho_Tj, rho_Tgs)
R["4c"] = (w_optimo,rho_omega_min)   # (rho_Tomega, omega_optimo)
R["4d"] = (x,r)   # (x_solve, res_inf)
```

```python
import numpy as np
from scipy.linalg import solve
from scipy.optimize import fsolve

A = np.array([[ 7, -2,  1],
              [-2,  8, -3],
              [ 1, -3,  9]], dtype=float)
b = np.array([20.0, -5.0, 32.0])

def F(v): # F(X) con X = [x, y]
    x, y  = v
    return np.array([x**2 + y**2 - 5, x*y-2])

##Punto 5a)
x_solve = solve(A,b)
r = A @ x_solve - b
r_inf = np.linalg.norm(r,np.inf)
det_a = np.linalg.det(A)

#Punto 5b)
cond_inf_a = np.linalg.cond(A, np.inf)

#Descomponer A ; A = D - L - U

D = np.diag(np.diag(A))
L = -np.tril(A,k = - 1)
U = -np.triu(A,k = 1)

Tj = np.linalg.inv(D) @ (L + U)
Tgs = np.linalg.inv(D - L) @ U

rho_tj = max(abs(np.linalg.eigvals(Tj)))
rho_tgs = max(abs(np.linalg.eigvals(Tgs)))


def diagonal_dominante(A):
    diag = np.abs(np.diag(A))
    suma_diagonal  = np.sum(np.abs(A),axis = 1) - diag
    es_diagonal_dominante = bool(np.all(diag > suma_diagonal))
    return es_diagonal_dominante

A_diag_dom = diagonal_dominante(A)

#Punto 5c)
b_modif = np.array([20.1, -5.0, 32.0])
x_solve_modif = solve(A,b_modif)
r_modif = A @ x_solve - b
r_inf_modif = np.linalg.norm(r,np.inf)

#Punto 5d)

f_1 = F([1, 1])

x_0 = [1.5 , 1.5]
x_1 = [2.5 , 0.5]
x_2 = [-1.5 , -1.5]
x_3 = [0.5 , 2.5]

valor_inicial  = [x_0,x_1,x_2,x_3]
soluciones = []


for punto in valor_inicial:
    sol  = fsolve(F,punto)
    soluciones.append(np.round(sol))

sol_distintas = np.unique(soluciones,axis = 0)#Esto no esta funcionando



R["5a"] = (x_solve,r_inf,det_a)   # (x_solve, res_inf, det_A)
R["5b"] = (cond_inf_a,A_diag_dom,rho_tj,rho_tgs)   # (cond_inf_A, es_diagonal_dominante, rho_Tj, rho_Tgs)
R["5c"] = (x_solve_modif,abs(r_inf - r_inf_modif))   # (x_perturbada, dif_inf)
R["5d"] = (f_1,soluciones,sol_distintas)   # (F(1,1), lista de las 4 soluciones, numero de soluciones distintas)
```
