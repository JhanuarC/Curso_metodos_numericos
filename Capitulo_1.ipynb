# Introducción al Método de la Bisección: Una Perspectiva Académica

**Bienvenidos, estudiantes.** Como su profesor de Métodos Numéricos, hoy vamos a adentrarnos en uno de los algoritmos más fundamentales y robustos que existen para el análisis numérico: **el Método de la Biseción**.

En el contexto de las ingenierías y ciencias aplicadas, casi siempre nos enfrentamos al problema de encontrar una solución exacta a una ecuación no lineal. Es decir, queremos hallar un valor $x$ tal que:
$$ f(x) = 0 $$

A menudo, estas funciones son tan complejas (cúbicas, trascendentes o derivadas de sistemas físicos) que no podemos despejar $x$ algebraicamente. Aquí es donde entran los **Métodos Numéricos**.

---

## ¿Por qué el Método de la Biseción?

Imaginen una función continua dibujada en un plano cartesiano. Si sabemos que la curva cruza el eje horizontal (el eje X) entre dos puntos $a$ y $b$, pero no sabemos exactamente dónde, necesitamos una estrategia para "acercarnos" a ese punto de cruce.

El Método de la Biseción es la implementación computacional del **Teorema del Valor Intermedio**. Es el método más intuitivo porque imita lo que haríamos si quisiéramos encontrar un objeto escondido en una habitación oscura: dividimos el espacio por la mitad y verificamos dónde está.

### Fundamento Matemático
Para que este método funcione, debemos cumplir dos condiciones previas (que son vitales para entender su teoría):

1.  **Continuidad:** La función $f(x)$ debe ser continua en el intervalo cerrado $[a, b]$. Si hay un salto o una asíntota vertical dentro del intervalo, la biseción puede fallar.
2.  **Cambio de Signo (Teorema del Valor Intermedio):** Debe cumplirse que:
    $$ f(a) \cdot f(b) < 0 $$
    Esto significa que $f(a)$ y $f(b)$ tienen signos opuestos (uno positivo, otro negativo). Por lo tanto, la gráfica debe cruzar el eje X al menos una vez dentro de ese intervalo.

---

## ¿Cómo funciona? (Intuición del Algoritmo)

El algoritmo es un proceso iterativo de **reducción de incertidumbre**:

1.  Calculamos el punto medio $c = \frac{a + b}{2}$.
2.  Evaluamos la función en ese punto: $f(c)$.
3.  Si $f(c)$ es cero, ¡tenemos la raíz exacta!
4.  Si no, evaluamos si $f(a)$ y $f(c)$ tienen signos opuestos.
    *   **Sí:** La raíz está en el intervalo izquierdo $[a, c]$. Actualizamos $b = c$.
    *   **No:** La raíz debe estar en el lado derecho $[c, b]$. Actualizamos $a = c$.

En cada iteración, reducimos la incertidumbre (el ancho del intervalo) a la mitad. Esto garantiza que el error absoluto se reduce exponencialmente con cada paso.

---

## Análisis de Convergencia y Error

Para un estudiante universitario, no basta con saber *cómo* funciona, sino *cuán rápido*.

El error absoluto en la iteración $n$ está acotado por:
$$ |x_{true} - x_n| \leq \frac{b_0 - a_0}{2^n} $$
Donde $(b_0 - a_0)$ es el ancho inicial del intervalo.

Esto nos indica que la **convergencia es lineal** con una tasa de aproximadamente $0.5$ por iteración. Es decir, cada paso reduce el error en un 50%. No es tan rápido como métodos como Newton-Raphson (que tienen convergencia cuadrática), pero es **garantizado**.

### Ventajas y Desventajas

| Característica | Descripción |
| :--- | :--- |
| **Ventaja Principal** | Es muy robusto. Siempre converge si se cumplen las condiciones iniciales, incluso con una estimación inicial "mala". |
| **Desventaja Principal** | Convergencia lenta (lineal). Puede requerir muchas iteraciones para alta precisión. |
| **Requisitos** | Necesita evaluar la función en dos puntos y verificar continuidad. No necesita derivadas ($f'(x)$). |

---

## ¿Cuándo usarlo?

En un flujo de trabajo real, el Método de la Biseción suele usarse como:
1.  **Prueba inicial:** Para confirmar que existe una raíz en un rango antes de intentar métodos más rápidos pero inestables (como Newton-Raphson).
2.  **Problemas con derivadas difíciles:** Cuando calcular $f'(x)$ es costoso o complejo, la biseción evita ese costo computacional.


```python
#Código para una funcion 

f = lambda x: x**3 + 2*x - 5

#Encontremos su raíz usando el método de la bisección


def biseccion(f, a, b, tol):
    if f(a) * f(b) >= 0:
        print("El método de bisección no puede aplicarse.")
        return None

    ciclos = 0
    while (b - a) / 2.0 > tol:
        ciclos = ciclos + 1
        c = (a + b) / 2.0
        if f(c) == 0:
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c
    print(f"Número de ciclos: {ciclos}")
    return (a + b) / 2.0

# Definimos el intervalo [a, b] y la tolerancia
a = 1
b = 3
tol = 1e-5

# Llamamos a la función de bisección
raiz = biseccion(f, a, b, tol)
print(f"La raíz aproximada es: {raiz}")
```

## Convergencia y Estimación A Priori

## TEOREMA DE CONVERGENCIA

El método de biseción genera una sucesión que converge a la raíz $p$. El error absoluto en la iteración $n$ está estrictamente acotado por la mitad de la longitud del intervalo en ese paso:

$$ |p_n - p| \leq \frac{b-a}{2^n}, \quad \forall n \geq 1 $$

Por lo tanto, el número de iteraciones necesarias para asegurar una tolerancia $\epsilon$ se puede pre-calcular resolviendo la desigualdad:

$$ n \geq \frac{\log\left(\frac{b-a}{\epsilon}\right)}{\log(2)} $$


# Método de Punto Fijo (Iteración de Punto Fijo)

El **Método del Punto Fijo** es una técnica numérica iterativa utilizada para resolver ecuaciones de la forma $f(x) = 0$. Consiste en transformar el problema de hallar una raíz en el problema equivalente de encontrar un **punto fijo** de una función auxiliar $g(x)$.

---

## 1. Definición Teórica

* **Punto Fijo:** Un número $\alpha$ es un *punto fijo* de una función $g(x)$ si al evaluarse en la función devuelve el mismo valor:
  $$g(\alpha) = \alpha$$
* **Relación con Raíces:** Dada la ecuación $f(x) = 0$, se reordena algebraicamente para aislar una $x$:
  $$f(x) = 0 \iff x = g(x)$$
  Si $\alpha$ es raíz de $f(x)$, también es punto fijo de $g(x)$.

---

## 2. Algoritmo Iterativo

Para encontrar la solución empezando desde un valor inicial $x_0$:

1. Formular la ecuación en forma $x = g(x)$.
2. Elegir un valor inicial $x_0$.
3. Generar la sucesión de aproximaciones mediante la fórmula recursiva:
   $$x_{k+1} = g(x_k), \quad k = 0, 1, 2, \dots$$
4. Criterio de parada: Detener el proceso cuando la diferencia absoluta entre dos aproximaciones consecutivas sea menor que la tolerancia especificada ($\epsilon$):
   $$|x_{k+1} - x_k| < \epsilon$$

---

## 3. Criterio de Convergencia

No toda función $g(x)$ generada algebraicamente garantizará llegar a la solución. 

> **Teorema de Convergencia:** Sea $\alpha$ un punto fijo de $g(x)$. El método convergerá a $\alpha$ en un intervalo cercano si la función es derivable y cumple:
> $$|g'(\alpha)| < 1$$

* **Si $|g'(\alpha)| < 1$:** El método **converge** (se acerca a la solución).
* **Si $|g'(\alpha)| > 1$:** El método **diverge** (se aleja de la solución).

---

## 4. Ejemplo Práctico Step-by-Step

Resolver la ecuación:
$$f(x) = x^2 - x - 2 = 0$$

### Paso 1: Transformar a $x = g(x)$
Despejando $x$:
$$x^2 = x + 2 \implies x = \sqrt{x + 2}$$
Definimos:
$$g(x) = \sqrt{x + 2}$$

### Paso 2: Verificar la convergencia
Calculamos la derivada $g'(x)$:
$$g'(x) = \frac{1}{2\sqrt{x + 2}}$$
Evaluamos cerca de la solución probable (ej. $x = 2$):
$$|g'(2)| = \left|\frac{1}{2\sqrt{4}}\right| = \frac{1}{4} = 0.25 < 1 \quad \text{(Converge)}$$

### Paso 3: Iterar
Usando valor inicial $x_0 = 1$ y tolerancia $\epsilon = 0.001$:

| $k$ | $x_k$ | $x_{k+1} = g(x_k)$ | Error: $|x_{k+1} - x_k|$ |
|---|---|---|---|
| 0 | 1.0000 | $\sqrt{1+2} = 1.7321$ | $0.7321$ |
| 1 | 1.7321 | $\sqrt{1.7321+2} = 1.9319$ | $0.1998$ |
| 2 | 1.9319 | $\sqrt{1.9319+2} = 1.9829$ | $0.0510$ |
| 3 | 1.9829 | $\sqrt{1.9829+2} = 1.9957$ | $0.0128$ |
| 4 | 1.9957 | $\sqrt{1.9957+2} = 1.9989$ | $0.0032$ |
| 5 | 1.9989 | $\sqrt{1.9989+2} = 1.9997$ | **$0.0008$** (< 0.001) |

**Resultado:** La raíz aproximada es **$x \approx 1.9997$** (valor exacto $= 2$).21

```python
#Metodo del punto fijo para encontrar la raíz de funciones
#Usando sympy
import sympy as sp


def iteracion_punto_fijo(g_func, x_actual, error=1e-4):
    x_nuevo = g_func(x_actual)
    
    if abs(x_nuevo - x_actual) < error:
        print(f"El cero de la función está cerca de {x_nuevo}")
        return x_nuevo
    else:

        return iteracion_punto_fijo(g_func, x_nuevo, error)


x  = sp.symbols('x')
#Definimos la función
f = x**2  - x - 2

# Definimos g(x) despejando x 
g = sp.sqrt(x + 2)

#Convertimos g auna funcion ejecutable de Python
g_func = sp.lambdify(x, g)

g_diff = sp.diff(g)
g_diff = sp.lambdify(x,g_diff)


if abs(g_diff(2)) < 1:
    print("Converge")
    print(g_func(0))
    iteracion_punto_fijo(g_func,1)
```

# Método de Newton-Raphson (Método de las Tangentes)

El **Método de Newton-Raphson** es un algoritmo numérico abierto e iterativo utilizado para encontrar las raíces o ceros de una función continua $f(x) = 0$. Utiliza la derivada de la función para acelerar la convergencia hacia la solución.

---

## 1. Interpretación Geométrica

1. Se parte de una estimación inicial $x_0$.
2. Se evalúa la función en $(x_0, f(x_0))$ y se traza la **recta tangente** a la curva en ese punto.
3. El punto donde esta recta tangente **corta el eje $x$** se toma como la siguiente aproximación ($x_1$).
4. Se repite el proceso iterativamente acercándose a la raíz exacta.

---

## 2. Deducción de la Fórmula Iterativa

La ecuación de la recta tangente en el punto $(x_k, f(x_k))$ es:

$$y - f(x_k) = f'(x_k)(x - x_k)$$

Buscamos la intersección con el eje horizontal ($y = 0$) y llamamos a la nueva posición $x_{k+1}$:

$$0 - f(x_k) = f'(x_k)(x_{k+1} - x_k)$$

Despejando $x_{k+1}$, obtenemos la **fórmula del método**:

$$x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}$$

---

## 3. Algoritmo Paso a Paso

1. Definir la función $f(x)$ y su derivada $f'(x)$.
2. Seleccionar un valor inicial $x_0$.
3. Calcular las iteraciones consecutivas:
   $$x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}$$
4. Criterio de parada: Detener cuando el error $\vert{}x_{k+1} - x_k\vert{} < \epsilon$ o $\vert{}f(x_{k+1})\vert{} < \epsilon$.

---

## 4. Pros y Contras

| Ventajas | Desventajas / Fallos |
| :--- | :--- |
| **Convergencia cuadrática:** Cerca de la raíz, el número de decimales correctos se duplica aproximadamente en cada iteración. | Si $f'(x_k) = 0$, se produce una división por cero y el método falla. |
| Requiere pocas iteraciones comparado con Bisección o Punto Fijo. | Sensible al valor inicial $x_0$: si está lejos de la raíz, puede divergir. |

---



```python

import sympy as sp

# Declaramos la variable matemática
x = sp.Symbol('x')

# Definimos la función f(x)
f_expr = x**2 - 2

# Calculamos la derivada analítica f'(x)
df_expr = sp.diff(f_expr, x)

# Convertimos a funciones ejecutables de Python
f = sp.lambdify(x, f_expr)
df = sp.lambdify(x, df_expr)

def newton_raphson(x0, tol=1e-5, max_iter=100):
    xk = x0
    for k in range(max_iter):
        if df(xk) == 0:
            print("Error: Derivada cero.")
            return None
        
        xk_next = xk - f(xk) / df(xk)
        
        if abs(xk_next - xk) < tol:
            print(f"Convergencia en la iteración {k+1}")
            return xk_next
            
        xk = xk_next
        
    return xk

# Ejemplo de uso: calcular sqrt(2) iniciando en x0 = 1.0
raiz = newton_raphson(x0=1.0)
print(f"Raíz encontrada: {raiz}")
```

# ¿Qué es un cero múltiple?

En el mundo de las funciones, no todas las raíces son iguales[cite: 3]. Algunas funciones simplemente "cruzan" el eje $x$, mientras que otras parecen "rebotar" o deslizarse suavemente sobre él[cite: 3].

*   **Cero Simple:** Es un corte directo[cite: 3]. Gráficamente, la función *atraviesa* el eje $x$ con una pendiente distinta de cero[cite: 3]. La función cambia de signo al cruzar $p$[cite: 3].
    $$f(x) = (x - p)$$
*   **Cero Múltiple:** Ocurre cuando la función es tangente al eje $x$[cite: 3]. Se dice que la raíz se "repite" $m$ veces[cite: 3]. La función puede tocar el eje sin cruzarlo[cite: 3].
    $$f(x) = (x - p)^m, \quad m \geq 2$$

> **💡 Intuición Gráfica:** Si la función presenta un mínimo o máximo local justo sobre el eje $x$, es un cero de multiplicidad *par* (la función rebota, $m = 2$)[cite: 3]. Si presenta un cambio de concavidad en el eje, es un cero de multiplicidad *impar* (la función se aplana pero cruza, $m = 3$)[cite: 3].

### Definición Formal
Una solución $p$ de $f(x) = 0$ es un **cero de multiplicidad $m$** de $f$ si para $x \neq p$ se puede escribir[cite: 3]:

$$f(x) = (x - p)^m q(x), \quad \lim_{x \to p} q(x) \neq 0$$

La función $q$ "absorbe" todo el comportamiento no-nulo cerca de $p$[cite: 3]. 

---

## Caracterización Analítica de la Multiplicidad

Para identificar la multiplicidad de una raíz sin graficar, usamos sus derivadas[cite: 4].

### Teorema de Multiplicidad
La función $f \in C^m[a, b]$ tiene un cero de multiplicidad $m$ en $p \in [a, b]$ si y solo si[cite: 4]:

$$0 = f(p) = f'(p) = f''(p) = \cdots = f^{(m-1)}(p)$$

y la derivada de orden $m$ es distinta de cero[cite: 4]:

$$f^{(m)}(p) \neq 0$$

Las primeras $m - 1$ derivadas se anulan en $p$; la $m$-ésima no[cite: 4].

---

## Convergencia del Método de Newton

### Newton converge cuadráticamente para ceros simples
Si $f \in C^2[a, b]$, $f(p) = 0$ y $f'(p) \neq 0$, la expansión de Taylor de segundo orden de $f$ alrededor de $p_n$ evaluada en $p$ da[cite: 5]:

$$0 = f(p_n) + f'(p_n)(p - p_n) + \frac{f''(\xi_n)}{2}(p - p_n)^2$$

Dividiendo entre $f'(p_n) \neq 0$ y despejando el error $p - p_{n+1}$[cite: 5]:

$$\lim_{n \to \infty} \frac{|p - p_{n+1}|}{|p - p_n|^2} = \frac{1}{2} \left| \frac{f''(p)}{f'(p)} \right|$$

Esto muestra convergencia de **orden $\alpha = 2$ (cuadrática)**[cite: 5].

### Newton converge linealmente para ceros múltiples
Para un cero múltiple de multiplicidad $m$, ocurre que $f'(p) = 0$[cite: 5]. Aunque el cociente $\frac{f(p_n)}{f'(p_n)}$ puede calcularse (el numerador se anula más rápido que el denominador), la convergencia se degrada[cite: 5]:

$$\lim_{n \to \infty} \frac{|p - p_{n+1}|}{|p - p_n|} = \frac{m - 1}{m} < 1$$

Esto es convergencia **lineal ($\alpha = 1$)**[cite: 5]. Para $m = 2$ la constante asintótica es $\frac{1}{2}$ (el error se reduce a la mitad en cada paso); para $m = 3$, es $\frac{2}{3}$[cite: 5].

> **🔑 Conclusión clave:** Para ceros simples Newton reduce el error *al cuadrado* en cada iteración[cite: 5]. Para ceros múltiples solo lo multiplica por la constante $\frac{m-1}{m}$[cite: 5]. La diferencia práctica es enorme: Newton estándar puede tomar 20+ iteraciones donde los métodos pro necesitan 3 o 4[cite: 5].

---

## Métodos para Ceros Múltiples

Cuando sabemos (o sospechamos) que hay una raíz múltiple, podemos recuperar la convergencia cuadrática[cite: 6]. Existen dos variantes:

### 1. Newton Acelerado
Se multiplica el paso de corrección por la multiplicidad $m$[cite: 6]:

$$p_{n+1} = p_n - m \cdot \frac{f(p_n)}{f'(p_n)}$$

*   ✅ Simple de implementar[cite: 6].
*   ⚠️ Requiere conocer $m$ de antemano[cite: 6].
*   Converge cuadráticamente si $m$ es el valor correcto[cite: 6].

### 2. Newton Modificado
Aplica Newton a $\mu(x) = \frac{f(x)}{f'(x)}$[cite: 6]:

$$p_{n+1} = p_n - \frac{f(p_n) f'(p_n)}{[f'(p_n)]^2 - f(p_n) f''(p_n)}$$

*   ✅ No requiere conocer $m$[cite: 6].
*   ⚠️ Requiere calcular $f''$[cite: 6].

**¿Por qué funciona el Newton Modificado?**
Si $f(x) = (x - p)^m q(x)$ con $q(p) \neq 0$, entonces $f'(x) = (x - p)^m q'(x) + m(x - p)^{m-1} q(x)$, y[cite: 6]:

$$\mu(x) = \frac{f(x)}{f'(x)} = (x - p) \cdot \frac{q(x)}{(x - p)q'(x) + m \, q(x)}$$

Como $\lim_{x \to p} \frac{q(x)}{(x - p)q'(x) + m \, q(x)} = \frac{1}{m} \neq 0$, la función $\mu$ tiene un **cero simple** en $p$[cite: 6]. Al aplicar Newton a $\mu$ se obtiene la fórmula del método modificado, y la convergencia es cuadrática[cite: 6].

### Resumen Comparativo

| Método | Convergencia | Requiere $m$ | Requiere $f''$ |
| :--- | :--- | :--- | :--- |
| **Newton estándar** | Lineal ($\alpha = 1$) | No | No |
| **Newton acelerado** | Cuadrática ($\alpha = 2$) | **Sí** | No |
| **Newton modificado** | Cuadrática ($\alpha = 2$) | No | **Sí** |
