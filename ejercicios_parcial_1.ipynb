```python
import numpy as np
import matplotlib.pyplot as plt
import scipy


h = lambda x: np.cos(5*x) * (np.exp(x) - np.arctan(x)) + np.exp(-x)

xx = np.linspace(-1,6,1000)##Dominio
yy = h(xx)##Rango

plt.plot(xx,yy)
plt.grid()
plt.plot()

plt.axis([-1,6,-1,1])#Nos da zoom
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import sympy as sp

h = lambda x: np.exp(x)*np.cos(3*x) - np.sin(5*x + 1)

#x = sp.symbols("x")

#f = sp.exp(x)*sp.cos(3*x) - sp.sin(5*x + 1)
##Encontrar ceros

xx = np.linspace(-2,1.5,100)
yy = h(xx)

plt.plot(xx,yy)

#plt.axis([-2,1.5,-1,1])#Nos da zoom
#plt.show()

#df = sp.diff(f)

df = lambda x: np.exp(x)*np.cos(3*x)-np.exp(x)*np.sin(3*x)*3-np.cos(5*x+1)*5 

yyd = df(xx)

plt.plot(xx,yyd, color = "green")
plt.grid()
```

```python
from scipy.optimize import bisect

raiz = bisect(df,a =-2,b = -1.5,xtol=1e-8,maxiter=100)
print(round(raiz,6))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fixed_point

h = lambda x:  x - (1/4)*np.cos(4*x) + (3/2)*np.exp(-x**2)*np.arctan(x)

xx = np.linspace(-1,4,1000)

yy = h(xx)

plt.plot(xx,yy)
plt.plot(xx,xx)
plt.grid()
plt.show()

pf = fixed_point(h, x0 = 0, xtol=1e-8,maxiter=500)
print(pf)

##La funcion tiene 5 puntos fijos
```

```python
from scipy.optimize import newton


dh = lambda x: (-2*np.exp(-2*x)*np.sin(3*x) + np.cos(3*x)*3*np.exp(-2*x)) + 100*np.sin(2*x)*2
puntos = [-4,-3,-2,0]

for punto in puntos:
    new = newton(h,x0 = punto,fprime =dh)
    print(f"Raiz en punto inicial: {punto}, el valor de la raíz es {new}")
```
